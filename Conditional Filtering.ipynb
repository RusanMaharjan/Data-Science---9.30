```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
# Find details of employees whose salary is less than 8000.
emp_salary = df[df['SALARY'] < 8000]
# emp_salary
```

```python
## Find details of employees whose salary is in range of 8000 to 17000.
emp_sal_range = df[(df['SALARY'] > 8000) & (df['SALARY'] < 17000)]
# emp_sal_range
```

```python
emp_sal_between = df[df['SALARY'].between(8000, 17000)]
# emp_sal_between
```

```python
## Find details of employees who works on Job AD_VP and IT_PROG.
emp_job = df[(df['JOB_ID'] == 'AD_VP') | (df['JOB_ID'] == 'IT_PROG')]
# emp_job
```

```python
emp_job_in = df[df['JOB_ID'].isin(['AD_VP', 'IT_PROG'])]
# emp_job_in
```

```python
## Find details of employees who works in job SA_MAN and SA_REP and whosee salary must be in range of 7000 to 12000.
job_sal = df[
    ((df['JOB_ID'] == 'SA_REP') | (df['JOB_ID'] == 'SA_MAN')) &
    ((df['SALARY'] >= 7000) & (df['SALARY'] <= 12000))
]
# job_sal
```
