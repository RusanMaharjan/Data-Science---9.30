# Machine Learning

```python
'''
Types of Machine Learning
-----------------------------
1. Supervised Learning
    - If our data has labelled data.
    - Supervised learning must have both feature and target.

    Types of Supervised Learning
    ----------------------------
    - Regression
        -> When our target value is continuous data.

        Models
        --------
        a. Linear Regression
        b. Random Forest Regressor
        c. Decision Tree Regressor
        d. XGBoost Regressor
        e. K Nearest Neighbor Regressor
        f. Support Vector Regressor
        
    - Classification
        -> When our target value is discrete / categorical.

        Models
        ---------
        a. Logisitic Regression
        b. Random Forest Classifier
        c. Decision Tree Classifier
        d. XGBoost Classifier
        e. K Nearest Neighbor Classifier
        f. Support Vector Classifier

2. Unsupervised Learning
    - If our data does not have labelled data.
    - Unsupervised learning only have features.

    Types of Unsupervised Learning
    -------------------------------
    - Clustering

    Models
    -------
        a. K Means Clustering
        b. DBSCAN Clustering
        c. Agglomerative Clustering
        d. Hierarchical Clustering
        e. PCA (Principle Component Analysis)
    
 -- 3. Reinforcement Learning

Note: Features must be of multi-dimension array and target must be of one-dimension array.

-----------------------------
Process of Machine Learning
-----------------------------
1. Data Wrangling, Data Analysis, Data Visualization
2. Separate Feature and Target
3. Train Test Split
    - X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y, test_size=0.2, random_state=42
      )

    - 80/20 Rule -> 80% Training and 20% Testing
    - Training Data -> Data from which model will learn patterns.
    - 70/30, 60/40, 50/50

4. Data Scaling
    -> To handle biasness of data.
    a. Feature Scaling
        -> To get data into equal set of range.
        
        - Standard Scaler (Standarization)
            -> If row of data is not in equal set of range.
            -> (Xi - mean) / sd
            -> Best Range (-1 to 1)
            -> Considerable Range (-3 to 3)

        - Min Max Scaler (Normalization)
            -> If row of data is in equal set of range.
            -> (Xi - min_val) / (max_val - min_val)
            -> Range (1 - 0)

    b. Label Encoding
    -> To convert categorical data into numeric data.
    a. Manual Encoder
        - If our category can be compared (Ordinal Data)
            - HE(4), SE(3), 5-9(2), PE(1), No (0)
            - Dictionary Mapping
            
    b. Label Encoder
        - If our category cannot be compared (Nominal Data)
        - LabelEncoder()
        - Sort Data in ascending order.
        - Gives each data a unique number starting from 0.
        - Yes, No (No(0), Yes(1))

    c. One-Hot Encoder (Binary Data)
        - If our category cannot be compared (Nominal Data)
        - OneHotEncoder()
        - Sort Data in ascending order.
        - For each category it forms a new column.
        - Gender -> M, F
        - Female_col, Male_col

5. Define Model
6. Fit Data into Model (80% Training (X_train), 80% Target (Y_train))
7. Predict Data (20% Testing) -> Answer
8. Metrics
    - Comparision between Y_pred (Predicted Data) and Actual Data (Y_test)
    a. Regression
        - MAE (Mean Absolute Error)
        - MSE (Mean Squared Error)
        - RMSE (Root Mean Squared Error) -> Loss Function
        - R2 Score (Residual Score) -> Cost Function


    b. Classification
        - Classification Report
        - Confusion Matrix

    c. Clustering
        - Silhouette Score
'''
```
