```python
'''
Data Cleaning
--------------
1. Fill or Remove null data
    a. Fill Null Data
        - fillna()

        i. Continuous Data
            - Mean
            - Median

            Rules
            ----------
            1. If mean and median value both are equal or almost equal then fill using mean value.
            2. Skewness
                - Positive or Right Skewness -> 0.5 and more -> Fill Median Value
                - Normal Distribution or Bell Curve -> 0 to  0.5 -> Fill Mean Value
                - Negative or Left Skewness -> less than 0 -> Fill Median Value

        ii. Discrete or Categorical
            - Mode

        iii. Continuous / Discrete / Categorical
            - Custom Data

    b. Drop Null Data
        - dropna()
        
2. Remove Duplicate Data
    a. Find duplicate data
        - duplicated()

    b. Drop duplicate data
        - drop_duplicates()

3. Remove unwated data
    a. Unwanted Data (text, regular expressions, digits, or patterns...)
        - str.replace()
        - Use of regular expression package

    b. Whitepaces
        - str.strip()

4. Change data types if required
    a. astype()

5. Remove outliers
    - IQR (Inter Quartile Range) Method 
'''
```

### Commission Pct

```python
comm_mean = round(df['COMMISSION_PCT'].mean(), 2)
comm_median = df['COMMISSION_PCT'].median()
comm_skew = df['COMMISSION_PCT'].skew()
```

```python
print(f'Commission Mean: {comm_mean}')
print(f'Commission Median: {comm_median}')
print(f'Commission Skewness: {comm_skew:,.2f}')
```

```python
# df['COMMISSION_PCT'].fillna(comm_mean, inplace=True)
```

```python
df.fillna({'COMMISSION_PCT': comm_mean}, inplace=True)
```

```python
df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(comm_mean)
```

### Manager and Department

```python
mgr_mode = df['MANAGER_ID'].mode()[0]
dep_mode = df['DEPARTMENT_ID'].mode()[0]

print(f'Manager Mode: {mgr_mode}\n')
print(f'Department Mode: {dep_mode}')
```

```python
df.fillna({
            'MANAGER_ID': mgr_mode, 
            'DEPARTMENT_ID': dep_mode}, 
          inplace=True)
```

## Remove unwated data

### Remove whitespace

```python
object_columns = hr.select_dtypes(['object'])
object_columns = object_columns.drop(columns = ['DOB', 'DateofHire', 'DateofTermination',
                                               'LastPerformanceReview_Date']) 
```

```python
def findUniqueData(dataframe, list_columns):
    for col in list_columns:
        print(col)
        print('-'*20)
        unique_data = dataframe[col].unique()
        print(unique_data)
        print('\n')
```

```python
columns = ['Employee_Name', 'Position', 'State', 'Gender', 'MaritalDesc',
       'CitizenDesc', 'HispanicLatino', 'RaceDesc', 'TermReason',
       'EmploymentStatus', 'Department', 'ManagerName', 'RecruitmentSource',
       'PerformanceScore']

# findUniqueData(hr, columns)
```

```python
# hr['Department'] = hr['Department'].str.strip()
def removeWhitespace(dataframe, list_columns):
    for col in list_columns:
        dataframe[col] = dataframe[col].str.strip()
    print('Whitespace removed.')
```

```python
removeWhitespace(hr, columns)
```

### Remove unwanted text/characters

```python
'''
- replace(find, replace_value, regex=True/False)
'''
hr['Employee_Name'] = hr['Employee_Name'].str.replace(',', ' ', regex=False) 
```

### Change Data Type

```python
# astype(datatype)
hr['ManagerID'] = hr['ManagerID'].astype(int)
```

```python
# pandas function -> to_datetime(date column)
columns = ['DOB', 'DateofHire', 'DateofTermination', 'LastPerformanceReview_Date']

def convertDateTime(dataframe, columns):
    for col in columns:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed') 
    print('Datetime converted.')

# YYYY-MM-DD
# hr['DOB'] = pd.to_datetime(hr['DOB'], dayfirst=True)
hr['DOB'] = pd.to_datetime(hr['DOB'], format='mixed')
```

```python
convertDateTime(hr, columns)
```

## Outlier Detection and Removing

```python
salary_skew = hr['Salary'].skew()
print(f'Salary Skewness: {salary_skew}')
```

```python
q1 = hr['Salary'].quantile(0.25)
q3 = hr['Salary'].quantile(0.75)
```

```python
iqr = q3 - q1
iqr
```

```python
lower_fence = q1 - (1.5 * iqr)
upper_fence = q3 + (1.5 * iqr)
```

```python
hr = hr[(hr['Salary'] >= lower_fence) & (hr['Salary'] <= upper_fence)].copy()
```

```python
hr['Salary'].skew()
```

```python
hr.to_csv('cleaned_data.csv', index=False)
hr.to_excel('cleaned_data.xlsx', index=False)
hr.to_json('cleaned_data.json', indent=4)
```
