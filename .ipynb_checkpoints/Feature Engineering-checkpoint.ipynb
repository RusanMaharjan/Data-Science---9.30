```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_data.csv')
df.head()
```

```python
columns = ['DOB', 'DateofHire', 'DateofTermination', 'LastPerformanceReview_Date']

def convertDateTime(dataframe, columns):
    for col in columns:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed') 
    print('Datetime converted.')
```

```python
convertDateTime(df, columns)
```

## Extract Dates

```python
df['hire_year'] = df['DateofHire'].dt.year
df['hire_month'] = df['DateofHire'].dt.month
df['hire_day'] = df['DateofHire'].dt.day
```

```python
df['month_name'] = df['DateofHire'].dt.month_name()
df['day_name'] = df['DateofHire'].dt.day_name()
```

## Convert Continuous Data to Categorical Data

```python
def convertSalary(salary):
    if salary >= 0 and salary <= 60000:
        return 'Low Salary'
    elif salary > 60000 and salary <= 75000:
        return 'Average Salary'
    else:
        return 'High Salary'
```

```python
df['salary_label'] = df['Salary'].apply(convertSalary)
```

## Separate Data

```python
# split(find, n(n_of_columns) = index, expand=True/False(default))
```

```python
# df['salary_label']
```

```python
salary_split = df['salary_label'].str.split(' ', n=1, expand=True)
df['Sal1'] = salary_split[0]
df['Sal2'] = salary_split[1]
```

```python
df.to_csv('cleaned_data.csv', index=False)
```
