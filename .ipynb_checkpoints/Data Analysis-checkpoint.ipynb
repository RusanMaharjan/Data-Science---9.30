# Data Analysis

```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_data.csv')
df.head()
```

```python
'''
1. groupby
2. aggregate functions -> sum, mean, count, min, max, nunique
3. rename
4. reset_index()
5. agg()
'''
```

```python
# Find total employees per department
emp_dep = df.groupby('Department')['EmpID'].count()
# emp_dep
```

```python
emp_dep = df.groupby('Department')['EmpID'].count().reset_index().rename(
    columns = {'EmpID': 'Total Employees'}
)
# emp_dep
```

```python
emp_dep = df.groupby('Department')['EmpID'].count().rename_axis(
    'Departments').rename('Total Employees').to_markdown(
        numalign='center',
        stralign='center'
    )
print(emp_dep)
```

```python
def currencyConverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}'"{:,.2f}".format)
```

```python
# Display all details of employees department.
dep_details = df.groupby('Department').agg(
    Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Avg_Salary = ('Salary', 'mean'),
    Max_Salary = ('Salary', 'max'),
    Min_Salary = ('Salary', 'min')
).reset_index()

dep_details.columns = dep_details.columns.str.replace('_', ' ', regex=False)
# dep_details['Total Salary'] = dep_details['Total Salary'].map("Rs.{:,.2f}".format)
currencyConverter(dep_details, 'Rs.', 'Total Salary', 'Avg Salary', 'Max Salary', 'Min Salary')
dep_details
```

```python
dep_gen_details = df.groupby(['Department', 'Gender']).agg(Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Avg_Salary = ('Salary', 'mean'),
    Max_Salary = ('Salary', 'max'),
    Min_Salary = ('Salary', 'min')
).reset_index()

dep_gen_details.columns = dep_gen_details.columns.str.replace('_', ' ', regex=False)
# dep_details['Total Salary'] = dep_details['Total Salary'].map("Rs.{:,.2f}".format)
currencyConverter(dep_gen_details, 'Rs.', 'Total Salary', 'Avg Salary', 'Max Salary', 'Min Salary')

dep_gen_details
```

## KPIs (Key Performance Indicators)

```python
total_employees = df['EmpID'].count()
total_salary = df['Salary'].sum()
total_manager = df['ManagerID'].nunique()
total_department = df['Department'].nunique()
total_absence_days = df['Absences'].sum()
total_project = df['SpecialProjectsCount'].sum()

print(f'Total Employees: {total_employees}')
print(f'Total Salary: Rs.{total_salary:,.2f}')
print(f'Total Managers: {total_manager}')
print(f'Total Departments: {total_department}')
print(f'Total Absence Days: {total_absence_days}')
print(f'Total Projects: {total_project}')
```

## Time Series Analysis (Co-hort Analysis)

```python
df.columns
```

```python
'''
Growth % (MoM, YoY, QoQ, WoW)
- ((current_value - baseline_value)/baseline_value) * 100
'''
```

```python
## Find total number of hired employees in each year
emp_year = df.groupby('hire_year')['EmpID'].count().reset_index().rename(
    columns = {'hire_year': 'Hire Year', 'EmpID': 'Total Employees'}
)
emp_year['YoY(%)'] = ((emp_year['Total Employees'].pct_change().fillna(0))*100).map("{:,.2f}%".format)
emp_year
```

```python
## Find total number of hired employees in each month.
month_hire = df.groupby(['hire_month', 'month_name'])['EmpID'].count().reset_index().rename(
    columns = {'month_name': 'Month Name', 'EmpID': 'Total Employees'}
)[['Month Name', 'Total Employees']]
month_hire
```

```python
## Find total number of hired employees in each day.
weekdays = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
df['day_name'] = pd.Categorical(df['day_name'], categories=weekdays)
```

```python
day_hire = df.groupby(['day_name'], observed=False)['EmpID'].count().reset_index()
day_hire
```
