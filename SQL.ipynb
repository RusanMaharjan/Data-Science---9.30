```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
server = 'RUSAN'
database = 'BikeStores'

connection_url = f'mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes'
engine = create_engine(connection_url)
```

```python
query = """
    select
    	sc.customer_id, CONCAT(sc.first_name, ' ', sc.last_name) as customer_name, sc.email, sc.street, sc.city, sc.city,
    	sc.zip_code,
    	so.order_status, so.order_date,
    	case
    		when order_status = 1 then 'Pending'
    		when order_status = 2 then 'Processing'
    		when order_status = 3 then 'Rejected'
    		when order_status = 4 then 'Completed'
    	End as order_label
    from sales.customers sc
    join sales.orders so
    on sc.customer_id = so.customer_id
    where so.order_status = 3;
"""
customer_order = pd.read_sql(query, engine)
customer_order.head()
```
