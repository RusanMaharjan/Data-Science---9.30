```python
import pandas as pd
```

## Handle Duplicate Data

```python
duplicate_data = df[df.duplicated(subset='id', keep='first')]
duplicate_data
```

## Fill Null Values

```python
df.fillna({'emp_title': 'Unknown Tile'}, inplace=True)
```

```python
# Extract Month
df['month_num'] = df['issue_date'].dt.month
df['issue_month'] = df['issue_date'].dt.month_name()
```

```python
current_month = 12
previous_month = 11
```

## KPIs

### Total Loan Applications

```python
Total_Loan_Application = df['id'].count()
MTD_Total_Loan_Application = df[df['month_num'] == current_month]['id'].count()
PMTD_Total_Loan_Application = df[df['month_num'] == previous_month]['id'].count()
MoM_Total_Loan_Application = ((MTD_Total_Loan_Application - PMTD_Total_Loan_Application)/PMTD_Total_Loan_Application)*100

print(f'Total Loan Application: {Total_Loan_Application}')
print(f'MTD Total Loan Application: {MTD_Total_Loan_Application}')
print(f'PMTD Total Loan Application: {PMTD_Total_Loan_Application}')
print(f'MoM Total Loan Application: {MoM_Total_Loan_Application:,.2f}%')
```

### Total Funded Amount

```python
Total_Funded_Amount = df['loan_amount'].sum()
MTD_Total_Funded_Amount = df[df['month_num'] == current_month]['loan_amount'].sum()
PMTD_Total_Funded_Amount = df[df['month_num'] == previous_month]['loan_amount'].sum()
MoM_Total_Funded_Amount = ((MTD_Total_Funded_Amount - PMTD_Total_Funded_Amount)/PMTD_Total_Funded_Amount)*100

print(f'Total Funded Amount: Rs.{Total_Funded_Amount:,.2f}')
print(f'MTD Total Funded Amount: Rs.{MTD_Total_Funded_Amount:,.2f}')
print(f'PMTD Total Funded Amount: Rs.{PMTD_Total_Funded_Amount:,.2f}')
print(f'MoM Total Funded Amount: {MoM_Total_Funded_Amount:,.2f}%')
```

### Total Amount Received

```python
Total_Amount_Received = df['total_payment'].sum()
MTD_Total_Amount_Received = df[df['month_num'] == current_month]['total_payment'].sum()
PMTD_Total_Amount_Received = df[df['month_num'] == previous_month]['total_payment'].sum()
MoM_Total_Amount_Received = ((MTD_Total_Amount_Received - PMTD_Total_Amount_Received)/PMTD_Total_Amount_Received)*100

print(f'Total Amount Received: Rs.{Total_Amount_Received:,.2f}')
print(f'MTD Total Amount Received: Rs.{MTD_Total_Amount_Received:,.2f}')
print(f'PMTD Total Amount Received: Rs.{PMTD_Total_Amount_Received:,.2f}')
print(f'MoM Total Amount Received: {MoM_Total_Amount_Received:,.2f}%')
```

### Average Interest Rate

```python
Average_Interest_Rate = (df['int_rate'].mean())*100
MTD_Average_Interest_Rate = (df[df['month_num'] == current_month]['int_rate'].mean())*100
PMTD_Average_Interest_Rate = (df[df['month_num'] == previous_month]['int_rate'].mean())*100
MoM_Average_Interest_Rate = ((MTD_Average_Interest_Rate - PMTD_Average_Interest_Rate)/PMTD_Average_Interest_Rate)*100

print(f'Average Interest Rate: {Average_Interest_Rate:,.2f}%')
print(f'MTD Average Interest Rate: {MTD_Average_Interest_Rate:,.2f}%')
print(f'PMTD Average Interest Rate: {PMTD_Average_Interest_Rate:,.2f}%')
print(f'MoM Average Interest Rate: {MoM_Average_Interest_Rate:,.2f}%')
```

### Average Debt to Income Ratio

```python
Average_DTI = (df['dti'].mean())*100
MTD_Average_DTI = (df[df['month_num'] == current_month]['dti'].mean())*100
PMTD_Average_DTI = (df[df['month_num'] == previous_month]['dti'].mean())*100
MoM_Average_DTI = ((MTD_Average_DTI - PMTD_Average_DTI)/PMTD_Average_DTI)*100

print(f'Average DTI: {Average_DTI:,.2f}%')
print(f'MTD Average DTI: {MTD_Average_DTI:,.2f}%')
print(f'PMTD Average DTI: {PMTD_Average_DTI:,.2f}%')
print(f'MoM Average DTI: {MoM_Average_DTI:,.2f}%')
```

## Good Loan vs. Bad Loan

```python
good_loan = df[df['loan_status'].isin(['Current', 'Fully Paid'])]
bad_loan = df[df['loan_status'] == 'Charged Off']
```

### Good Loan KPIs

```python
Good_Loan_Application = good_loan['id'].count()
Good_Loan_Application_Pct = (Good_Loan_Application / Total_Loan_Application)*100
Good_Loan_Funded_Amount = good_loan['loan_amount'].sum()
Good_Loan_Received_Amount = good_loan['total_payment'].sum()

print(f'Good Loan Applications: {Good_Loan_Application}')
print(f'Good Loan Application (%): {Good_Loan_Application_Pct:,.2f}%')
print(f'Good Loan Funded Amount: Rs.{Good_Loan_Funded_Amount:,.2f}')
print(f'Good Loan Received Amount: Rs.{Good_Loan_Received_Amount:,.2f}')
```

### Bad Loan KPIs

```python
Bad_Loan_Application = bad_loan['id'].count()
Bad_Loan_Application_Pct = (Bad_Loan_Application / Total_Loan_Application)*100
Bad_Loan_Funded_Amount = bad_loan['loan_amount'].sum()
Bad_Loan_Received_Amount = bad_loan['total_payment'].sum()

print(f'Bad Loan Application: {Bad_Loan_Application}')
print(f'Bad Loan Application (%): {Bad_Loan_Application_Pct:,.2f}%')
print(f'Bad Loan Funded Amount: Rs.{Bad_Loan_Funded_Amount:,.2f}')
print(f'Bad Loan Received Amount: Rs.{Bad_Loan_Received_Amount:,.2f}')
```

## Loan Status Grid View

```python
def pctConverter(dataframe, suffix, *args):
    for col in args:
        dataframe[col] = (dataframe[col]*100).map("{:,.2f}"f'{suffix}'.format)

def currencyConverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f"{prefix}""{:,.2f}".format)
```

```python
mtd_data = df[df['month_num'] == current_month]
mtd_fra = mtd_data.groupby('loan_status').agg(
    MTD_Funded_Amount = ('loan_amount', 'sum'),
    MTD_Received_Amount = ('total_payment', 'sum')
).reset_index()
```

```python
loan_grid_view = df.groupby('loan_status').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum'),
    Average_Int_Rate = ('int_rate', 'mean'),
    Average_DTI = ('dti', 'mean')
).reset_index().rename(
    columns = {'loan_status': 'Loan Status'}
)

loan_grid_view['MTD_Funded_Amount'] = mtd_fra['MTD_Funded_Amount']
loan_grid_view['MTD_Received_Amount'] = mtd_fra['MTD_Received_Amount']

pctConverter(loan_grid_view, '%', 'Average_Int_Rate', 'Average_DTI')
currencyConverter(loan_grid_view, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount', 'MTD_Funded_Amount', 
                  'MTD_Received_Amount')

loan_grid_view.columns = loan_grid_view.columns.str.replace('_', ' ', regex=False)


loan_grid_view
```

## Analysis

```python
def financeAnalysis(dataframe, col_name, rename):
    data = dataframe.groupby(f'{col_name}').agg(
        Total_Loan_Applications = ('id', 'count'),
        Total_Funded_Amount = ('loan_amount', 'sum'),
        Total_Received_Amount = ('total_payment', 'sum')
    ).reset_index().rename(
        columns = {col_name : rename}
    )

    currencyConverter(data, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

    data.columns = data.columns.str.replace('_', ' ', regex=False)

    return data
```

### Monthly Trends

```python
monthly_trend = df.groupby(['month_num', 'issue_month']).agg(
    Total_Loan_Applications = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'issue_month': 'Issued Month'}
)[['Issued Month', 'Total_Loan_Applications', 'Total_Funded_Amount', 'Total_Received_Amount']]

currencyConverter(monthly_trend, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

monthly_trend.columns = monthly_trend.columns.str.replace('_', ' ', regex=False)

monthly_trend
```

### Regional Analysis

```python
regional_trend = df.groupby('address_state').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'address_state': 'Regional Address'}
)

currencyConverter(regional_trend, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

regional_trend.columns = regional_trend.columns.str.replace('_', ' ', regex=False)
regional_trend
```

### Loan Term Analysis

```python
loan_term_analysis = df.groupby('term').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'term': 'Loan Term'}
)

currencyConverter(loan_term_analysis, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

loan_term_analysis.columns = loan_term_analysis.columns.str.replace('_', ' ', regex=False)

loan_term_analysis
```

### Employee Length

```python
financeAnalysis(df, 'emp_length', 'Employee Length').sort_values(by='Total Loan Applications')
```

### Loan Purpose

```python
financeAnalysis(df, 'purpose', 'Loan Purpose')
```

### Home Ownership

```python
financeAnalysis(df, 'home_ownership', 'Home Ownership')
```
